```python
import pandas as pd

url_fake = "https://raw.githubusercontent.com/KaiDMML/FakeNewsNet/master/dataset/politifact_fake.csv"
url_real = "https://raw.githubusercontent.com/KaiDMML/FakeNewsNet/master/dataset/politifact_real.csv"

fake = pd.read_csv(url_fake)
real = pd.read_csv(url_real)

fake["label"] = 0
real["label"] = 1

df = pd.concat([fake, real], ignore_index=True)
print(df.shape)
print(df.head())
```

```python
print(df['label'].value_counts())
print("\nMissing values:")
print(df.isnull().sum())
```

```python
df = df.dropna(subset=['title'])
```

```python
X = df['title']
y = df['label']

print(f"Total samples: {len(df)}")
print(f"Fake: {sum(y==0)}, Real: {sum(y==1)}")
```
